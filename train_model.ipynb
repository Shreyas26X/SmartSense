```python
import pandas as pd
import numpy as np
from datetime import datetime
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
import joblib
```

```python
df = pd.read_csv("StyleSense_Dataset_updated.csv")  # change file name
df.head()
```

```python
# Use InvoiceDate if available
if 'InvoiceDate' in df.columns:
    df['Purchase_Date'] = pd.to_datetime(df['InvoiceDate'])
else:
    df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'])
```

```python
today = datetime.now()

# Recency
df['Recency'] = (today - df['Purchase_Date']).dt.days

# Total Price
if 'UnitPrice' in df.columns and 'Quantity' in df.columns:
    df['Total_Price'] = df['Quantity'] * df['UnitPrice']

# Avg Order Value
df['Avg_Order_Value'] = df['Total_Price'] / df['Quantity'].replace(0, 1)

# Clean data
df = df.dropna(subset=['Recency', 'Avg_Order_Value'])
```

```python
scaler = MinMaxScaler()
X = scaler.fit_transform(df[['Recency', 'Avg_Order_Value']])
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(X)

print("Centroids:")
print(kmeans.cluster_centers_)
```

```python
joblib.dump(kmeans, "kmeans_model.joblib")
joblib.dump(scaler, "scaler.joblib")

print("Model saved successfully")
```
